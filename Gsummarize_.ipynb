```python
!pip install google-genai
from google import genai
from google.colab import userdata
client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))
MODEL="gemini-3.8-flash"
def summarize_text(long_text):
    prompt=f"""
summarize the following text in 2-3 simple senteces.
text:"{long_text}"
summary:
"""
    response = client.models.generate_content(model=MODEL,contents=prompt,)
    return response.text.strip()

text="""Python is a popular programming language created by guido van rossum in 1991.It is known for its simple,readable system which makes it a great choice for beginners.python is widely used in web development,data structures,artificial intelligence,and automation.many top companies like netflix,google,andinstagram use python in the systems."""
result = summarize_text(text)
print("original length", len(text))
print("summary", result)
```
