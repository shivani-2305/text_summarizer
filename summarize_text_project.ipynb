```python
!pip install google-generativeai
import google.generativeai as genai
from google.colab import userdata
genai.configure(api_key=userdata.get("GEMINI_API_KEY"))
model=genai.GenerativeModel("gemini-flash-latest")

def summarize_text(long_text):
    prompt=f"""
summarize the following text in 2-3 simple senteces.
text:"{long_text}"
summary:
"""
    response=model.generate_content(prompt)
    return response.text

article="""
 Python is a popular programming language cAPI_KEYreated by guido van rossum in 1991.It is known for its simple,readable system which makes it a great choice for beginners.python is widely used in web development,data structures,artificial intelligence,and automation.many top companies like netflix,google,andinstagram use python in the systems.
 """
result=summarize_text(article)
print("original length",len(article))
print("summary",result)
```
